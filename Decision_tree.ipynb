```python
import pandas as pd
import numpy as np

# Recreating our 7-day Tennis dataset
data = {
    'Outlook': ['Sunny', 'Overcast', 'Rain', 'Rain', 'Sunny', 'Overcast', 'Sunny'],
    'Humidity': ['High', 'High', 'High', 'Normal', 'Normal', 'Normal', 'High'],
    'Windy': ['No', 'No', 'Yes', 'No', 'No', 'Yes', 'Yes'],
    'Play': ['No', 'Yes', 'No', 'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)
print("--- Our Dataset ---")
print(df)
```

```python
def calculate_gini(labels):
    """Calculates the Gini Impurity of a list/series of target labels."""
    total_samples = len(labels)
    if total_samples == 0:
        return 0

    # Count how many 'Yes' and 'No' we have
    counts = labels.value_counts()

    # Calculate probabilities squared: (p_yes^2) + (p_no^2)
    sum_squared_probs = 0
    for label_count in counts:
        prob = label_count / total_samples
        sum_squared_probs += prob ** 2

    # Gini = 1 - sum of squared probabilities
    return 1 - sum_squared_probs

# Let's test it on our starting data (4 'Yes', 3 'No')
root_gini = calculate_gini(df['Play'])
print(f"Starting (Root Node) Gini Impurity: {root_gini:.4f}")
```

```python
print("--- Simulating a split on 'Humidity' ---")

# 1. Split the data into two groups
high_humidity_group = df[df['Humidity'] == 'High']['Play']
normal_humidity_group = df[df['Humidity'] == 'Normal']['Play']

# 2. Calculate impurity for each group independently
gini_high = calculate_gini(high_humidity_group)
gini_normal = calculate_gini(normal_humidity_group)

print(f"High Humidity Group ({len(high_humidity_group)} samples): Gini = {gini_high:.4f}  (Labels: {list(high_humidity_group)})")
print(f"Normal Humidity Group ({len(normal_humidity_group)} samples): Gini = {gini_normal:.4f} (Labels: {list(normal_humidity_group)})")

# 3. Calculate the weighted average of the two groups based on their size
total_samples = len(df)
weighted_gini_after_split = (
    (len(high_humidity_group) / total_samples) * gini_high +
    (len(normal_humidity_group) / total_samples) * gini_normal
)

# 4. Calculate Gini Gain (Higher is better!)
gini_gain = root_gini - weighted_gini_after_split

print("\n--- Final Split Verdict ---")
print(f"Weighted Gini after split: {weighted_gini_after_split:.4f}")
print(f"Gini Gain (Impurity Reduction): {gini_gain:.4f}")
```
